```python
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
%matplotlib inline

words = open('names.txt','r').read().splitlines()

chars = sorted(list(set(''.join(words))))
char_to_int = {char: idx + 1 for idx, char in enumerate(chars)}
char_to_int['.'] = 0
int_to_char = {idx:char for char,idx in char_to_int.items()}

xs, ys = [], []
for w in words:
    chars = ['.'] + list(w) + ['.']
    for ch1, ch2 in zip(chars, chars[1:]):
        xs.append(char_to_int[ch1])
        ys.append(char_to_int[ch2])

xs = torch.tensor(xs)
ys = torch.tensor(ys)

x_count = xs.nelement() # 228146

g = torch.Generator().manual_seed(2147483647)
weights = torch.randn((27, 27), generator=g, requires_grad=True)
```

```python
for k in range(1):

    #forward pass
    x_encoded = F.one_hot(xs, num_classes=27).float()
    logits = x_encoded @ weights
    counts = logits.exp()
    probs = counts / counts.sum(1, keepdims=True)

    p_true = probs[torch.arange(x_count), ys]

    log_loss = p_true.log()
    nll = -log_loss
    loss = nll.mean()
    print(loss.item())

    weights.grad = None
    loss.backward()

    weights.data += -50 * weights.grad
```
